### **MAKE IMAGES READABLE AGAIN**


*   using OpenCV



```python
!pip install opencv-python
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img_path='/content/unreadimg.jpeg'
img=cv2.imread(img_path)
gray=cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray=cv2.resize(gray,(600,950))
_,result=cv2.threshold(gray,20,255, cv2.THRESH_BINARY)
adap=cv2.adaptiveThreshold(gray,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,11,2)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,3,1), plt.imshow(img[..., ::-1]), plt.title("Original Image")  # Convert BGR to RGB for Matplotlib
plt.subplot(1,3,2), plt.imshow(result, cmap='gray'), plt.title("Thresholded Image")
plt.subplot(1,3,3), plt.imshow(adap, cmap='gray'), plt.title("Adaptive Thresholding")
plt.show()
```
